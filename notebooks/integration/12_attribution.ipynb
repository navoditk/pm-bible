## PREDICT
A portfolio holds the exact same assets as its benchmark, just at
different weights — no stock-picking or bond-picking within any bucket,
purely different sizing decisions. Before calculating: should the
Brinson *selection* effect come out to exactly zero, roughly zero, or
could it still be nonzero?

## Formula (Brinson attribution)
`allocation = (w_p - w_b) * r_b`

`selection = w_b * (r_p - r_b)`

`interaction = (w_p - w_b) * (r_p - r_b)`

Identity: `sum(allocation + selection + interaction) = active_return`

```python
import numpy as np

# Portfolio and benchmark hold the same 3 segments, same realized returns
# per segment (rp == rb) - only the weights differ.
wp = np.array([0.5, 0.3, 0.2])
wb = np.array([0.4, 0.4, 0.2])
rp = np.array([0.08, 0.05, 0.03])
rb = np.array([0.08, 0.05, 0.03])  # identical to rp: no selection skill by construction

# MANUAL FIRST:
allocation = None
selection = None
interaction = None
print("allocation:", allocation, " sum:", None if allocation is None else allocation.sum())
print("selection:", selection)

portfolio_return = float(wp @ rp)
benchmark_return = float(wb @ rb)
active_return = portfolio_return - benchmark_return
print("active return:", active_return)

# CHECK (uncomment after your attempt):
# assert np.allclose(selection, 0.0) and np.allclose(interaction, 0.0), (
#     "with rp==rb per segment, only the weighting decision can differ - "
#     "selection and interaction must be exactly zero"
# )
# assert np.isclose(allocation.sum(), active_return)
```
