## Formula
`leverage_ratio = total_debt / EBITDA` (lower is safer)

`interest_coverage_ratio = EBITDA / interest_expense` (higher is safer)

`fixed_charge_coverage_ratio = EBITDA / (interest_expense + mandatory_principal_payments)`

```python
# MANUAL FIRST:
issuer_a_leverage = None   # total_debt=200, ebitda=100
issuer_a_coverage = None   # ebitda=100, interest_expense=10
issuer_b_leverage = None   # total_debt=400, ebitda=100
issuer_b_coverage = None   # ebitda=100, interest_expense=30

print("A:", issuer_a_leverage, issuer_a_coverage)
print("B:", issuer_b_leverage, issuer_b_coverage)

# CHECK (uncomment after your attempt):
# import numpy as np
# assert np.isclose(issuer_a_leverage, 2.0) and np.isclose(issuer_a_coverage, 10.0)
# assert np.isclose(issuer_b_leverage, 4.0) and np.isclose(issuer_b_coverage, 100/30)
# assert issuer_b_leverage > issuer_a_leverage and issuer_b_coverage < issuer_a_coverage
```
