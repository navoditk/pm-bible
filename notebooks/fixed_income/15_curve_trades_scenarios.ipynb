`units_leg2 = hedge_ratio(dv01_leg1, dv01_per_unit_leg2)`

so the combined position has ~zero net DV01 while still expressing a view
on curve *shape* (steepening/flattening), not the level of rates.


```python
# 2s10s steepener: long $10mm 2Y notes, short an offsetting DV01 in 10Y notes.
dv01_2y_per_100mm = 19_000   # $ per bp for $100mm face of 2Y notes
dv01_10y_per_100mm = 85_000  # $ per bp for $100mm face of 10Y notes

long_2y_face = 10_000_000
dv01_2y_leg = dv01_2y_per_100mm * (long_2y_face / 100_000_000)

# MANUAL FIRST:
# use hedge_ratio(dv01_2y_leg, dv01_per_unit) to size the offsetting 10Y
# face amount so the trade is DV01-neutral. dv01_per_unit is the 10Y DV01
# per $1 face (derive it from dv01_10y_per_100mm).
short_10y_face = None
print(short_10y_face)
```
