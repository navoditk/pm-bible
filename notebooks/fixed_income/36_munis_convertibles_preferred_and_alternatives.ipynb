## PREDICT (convertible bonds)
A $1,000-par convertible bond has a conversion ratio of 20 (20 shares
per bond) and trades at a price of 950. The stock trades at $45. Is the
bond trading above or below its immediate conversion value - and is
that "above" or "below" actually a sign of mispricing?

## Formula (conversion value and premium)
`conversion_value = conversion_ratio * stock_price`
`conversion_premium = (bond_price - conversion_value) / conversion_value`

```python
from pm.fixed_income.convertibles import conversion_premium, conversion_value

conversion_ratio, stock_price, bond_price = 20, 45.0, 950.0

# MANUAL FIRST:
cv = None  # conversion_value(conversion_ratio, stock_price)
cp = None  # conversion_premium(bond_price, cv)
print("conversion value:", cv, " conversion premium:", cp)

# CHECK (uncomment after your attempt):
# import numpy as np
# assert np.isclose(cv, 900.0)
# assert np.isclose(cp, 0.055555556, atol=1e-8)
```

## Was your PREDICT right?
The bond trades above its conversion value - that's normal, not a sign
of mispricing. It reflects the remaining time value of the embedded
option to convert later at a potentially better stock price, plus the
bond's downside floor if the stock falls. A premium near zero would mean
the option has almost no remaining time value - typically because the
stock is already trading well above the conversion price.
