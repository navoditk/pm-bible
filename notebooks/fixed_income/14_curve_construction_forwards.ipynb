## Formula (bootstrapping)
For consecutive annual-pay par bonds, solve shortest-to-longest maturity:

`100 = sum_{k=1}^{n-1} coupon / (1+z_k)^k + (coupon+100) / (1+z_n)^n`

where `z_1 .. z_{n-1}` are already known from earlier steps and you solve
for `z_n`.
