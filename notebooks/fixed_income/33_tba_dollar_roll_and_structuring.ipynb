## PREDICT (the roll)

next month at 100.625 (a 0.375 drop) on a 5% coupon pool. GC repo for
agency MBS is trading at 5.30%. Without calculating: is giving up a
month of 5% coupon income worth a 0.375-point drop, roughly? Is this roll
likely to come out cheaper or more expensive than GC repo?

## Formula
`implied_financing_rate = (coupon_income - drop_income) / near_amount / horizon_years`

where `coupon_income` and `drop_income` are dollar amounts over the roll
period (not annualized), `near_amount = near_price/100 * face`, and
`horizon_years` is the roll period in years (`1/12` for a standard
monthly roll).

```python
from pm.fixed_income.mbs import dollar_roll_implied_financing_rate

face = 1_000_000.0
coupon_rate = 0.05
near_price, far_price = 101.00, 100.625
horizon_years = 1 / 12

# MANUAL FIRST:
coupon_income = None   # face * coupon_rate / 12
drop_income = None     # (near_price - far_price) / 100 * face
near_amount = None     # near_price / 100 * face
implied_financing_rate = None  # use dollar_roll_implied_financing_rate

print("coupon_income:", coupon_income)
print("drop_income:", drop_income)
print("near_amount:", near_amount)
print("implied_financing_rate (%):", None if implied_financing_rate is None else implied_financing_rate * 100)

# CHECK (uncomment after your attempt):
# import numpy as np
# assert np.isclose(coupon_income, 4166.6667, atol=1e-2)
# assert np.isclose(drop_income, 3750.0)
# assert np.isclose(near_amount, 1_010_000.0)
# assert np.isclose(implied_financing_rate, 0.004950495, atol=1e-6)
```

## Was your PREDICT right?
At ~0.50% implied financing, this roll is far cheaper than the 5.30% GC
repo rate quoted above - deeply "special." That tells you this
coupon/program is in unusually high demand to be short for delivery, not
just that financing happens to be cheap this month.
