## PREDICT
All capital weights in a 4-asset portfolio are 25%. Must all risk shares
also be 25%? If one asset has much higher volatility and is positively
correlated with everything else, will its risk share be above or below
25%?

## Formula
`MCR_i = (Sigma w)_i / sigma_p`

`CR_i = w_i * MCR_i`

Invariant: `sum(CR_i) = sigma_p`

(asset 1 variance 0.04, asset 2 variance 0.01, zero covariance).

`CR = w * MCR`. Work this out by hand before the next cell — with zero
correlation, `Sigma @ w` is just `[w1*var1, w2*var2]`.
